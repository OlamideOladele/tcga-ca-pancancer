```python
# here we use the count function to count how many individual bases we have
seq = "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"
g_count = seq.count("G")
c_count = seq.count("C")
a_count = seq.count("A")
t_count = seq.count("T")
print(len(seq), g_count, c_count, a_count, t_count) #used len function to calculate how many bases we have in total
```

```python
total_length = len(seq)
# Purine is A + G
purine_ratio = a_count + g_count
# Pyrimidine is C + T
pyrimidine_ratio = c_count + t_count
total_ratio = round((purine_ratio/pyrimidine_ratio), 2)
print (f"The total length is: {total_length} and the purine to pyrimidinre ratio i: {total_ratio}")
```
